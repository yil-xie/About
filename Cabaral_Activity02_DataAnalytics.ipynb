#Predictive Analytics

```python
df["pass"] = (df["G3"]>=10).astype(int)
```

```python
print(df["pass"].value_counts)
```

```python
features = [
    'age',
    'studytime',
    'failures',
    'G1',
    'G2'
]

x = df[features]
y = df["pass"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
model = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
    )
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)

print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['fail','pass'])
disp.plot()
plt.show()

print("Student pass/fai l confusion matrix")
print(cm)
```

```python
results = x_test.copy()
results["actual"] = y_test
results["predicted"] = y_pred
results.head(10)
```

#Prescriptive Analytics

```python
results["Risk"] = results["predicted"].map({
    0: "At Risk",
    1: "Likely to Pass"
})

results.head(10)
```

```python
results["Recommendation"] = results["predicted"].map({
    0: "Prioritize for academic intervention",
    1: "Continue regular monitoring"
})

results.head(10)
```

```python
probabilities = model.predict_proba(x_test)

results["Probability_of_Failing"] = probabilities[:, 0]

results.sort_values(
    "Probability_of_Failing",
    ascending = False
).head(20)
```

```python
priority_students = results.sort_values(
    "Probability_of_Failing",
    ascending = False
).head(20)

priority_students.to_csv(
    "priority_students.csv",
    index=False
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
print(iris.feature_names)
```

```python
print(iris.target_names)
```

```python
df = pd.DataFrame(
    data=iris.data,
    columns=iris.feature_names
    )
df["species"] = iris.target_names[iris.target]
df.head()
```

```python
print('number pf rpws: ', df.shape[0])
print('number of columns', df.shape[1])
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df['sepal length (cm)'],
    df['sepal width (cm)'],
)

plt.xlabel('sepal length (cm)')
plt.ylabel('sepal width (cm)')
plt.title('sepal length vs sepal width')
plt.show()
```

```python
plt.figure(figsize=(8,6))

for species in df['species'].unique():

  subset = df[df['species']==species]

  plt.scatter(
      subset['sepal length (cm)'],
      subset['sepal width (cm)'],
      label=species
  )

  plt.xlabel('sepal length (cm)')
  plt.ylabel('sepal width (cm)')
  plt.title('Iris Species based on Sepal measurements')
  plt.legend()
  plt.show()
```

```python
plt.figure(figsize=(8,6))

for species in df['species'].unique():

  subset = df[df['species']==species]

  plt.scatter(
      subset['petal length (cm)'],
      subset['petal width (cm)'],
      label=species
  )

  plt.xlabel('petal length (cm)')
  plt.ylabel('petal width (cm)')
  plt.title('Iris Species based on Petal measurements')
  plt.legend()
  plt.show()
```

```python
x = iris.data
y = iris.target

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print('training observations: ', len(x_train))
print('testing observations: ', len(x_test))
```

```python
model = KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(x_train, y_train)
```

```python
y_prod = model.predict(x_test)
print(y_prod)
```

```python
print('actual: ')
print(y_test)

print('\nPredicted: ')
print(y_prod)
```

```python
accuracy = accuracy_score(y_test, y_prod)
print('accuracy: ', accuracy)
```

```python
new_flower = [[4.9, 3, 1.4, 0.2]]
```

```python


prediction = model.predict(new_flower)
print(
    "predicted species: ",
    iris.target_names[model.predict(new_flower)[0]]
)
```

```python
probabilities = model.predict_proba(new_flower)
print(probabilities)
```

```python
for species, probability in zip(
    iris.target_names,
    probabilities[0]
    ):

  print(f'{species}: {probability * 100:.2f}%')
```

```python
k_values = [1, 3, 5, 7, 9]
accuracies = []

for k in k_values:
  model = KNeighborsClassifier(n_neighbors=k)
  model.fit(x_train, y_train)
  y_pred = model.predict(x_test)
  accuracy = accuracy_score(y_test, y_pred)
  accuracies.append(accuracy)
  print(f"K={k}, Accuracy = {accuracy:2f}")
```

```python
plt.figure(figsize=(8,6))
plt.plot(
    k_values,
    accuracies,
    marker='o'
    )

plt.xlabel('K')
plt.ylabel('Accuracy')
plt.title('KNN Model Accuracy for Different K Values')
plt.show()
```
